```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("NotebookApp") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
print("Spark version:", spark.version)
```

```python
data = [
    (1, "Amit", "Engineering", "Kanpur", 75000, 28),
    (2, "Priya", "Marketing", "Delhi", 62000, 30),
    (3, "Rahul", "Engineering", "Lucknow", 82000, 26),
    (4, "Sneha", "HR", "Delhi", 48000, 29),
    (5, "Vikram", "Sales", "Mumbai", 55000, 35),
    (6, "Anjali", "Engineering", "Delhi", 91000, 32),
    (7, "Rohan", "Marketing", "Mumbai", 58000, 27),
    (8, "Kavya", "HR", "Kanpur", 46000, 24),
    (9, "Arjun", "Sales", "Lucknow", 52000, 31),
    (10, "Neha", "Engineering", "Mumbai", 88000, 29),
    (11, "Karan", "Sales", "Delhi", 61000, 38),
    (12, "Pooja", "Marketing", "Kanpur", 59000, 26)
]

columns = [
    "employee_id", "name", "department",
    "city", "salary", "age"
]

df = spark.createDataFrame(data, columns)

df.show()
```

```python
from pyspark.sql.functions import col
x = df.filter(col("salary")<0)
x.show()
```

```python
df.filter(col("name").isNull()).show()
```

```python
df.filter(col("salary")>70000).show()
```

```python
from pyspark.sql.functions import trim
df = df.withColumn(
    "name",
    trim(col("name"))
)

df.show()
```

```python
df = df.withColumn(
    "annual_salary",
    col("salary")*12
)

df.show()
```

```python
report_df =df.filter(col("salary")>60000)
report_df.show()
```

```python
from pyspark.sql.functions import count,avg,sum
department_report = report_df.groupBy("department").agg(
    count("*").alias("total_employee"),
    avg("salary").alias("average_salary"),
    sum("salary").alias("total_salary")
)

department_report.show()
```
