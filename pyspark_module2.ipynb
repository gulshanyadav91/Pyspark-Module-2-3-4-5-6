```python
import os
import sys

os.environ["PYSPARK_PYTHON"] = sys.executable
os.environ["PYSPARK_DRIVER_PYTHON"] = sys.executable

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("NotebookApp") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
print("Spark version:", spark.version)
```

```python
data = [
    (1,  "Amit",   "Engineering", "Kanpur",  75000, 28),
    (2,  "Priya",  "Marketing",   "Delhi",   62000, 30),
    (3,  "Rahul",  "Engineering", "Lucknow", 82000, 26),
    (4,  "Sneha",  "HR",          "Delhi",   48000, 29),
    (5,  "Vikram", "Sales",       "Mumbai",  55000, 35),
    (6,  "Anjali", "Engineering", "Delhi",   91000, 32),
    (7,  "Rohan",  "Marketing",   "Mumbai",  58000, 27),
    (8,  "Kavya",  "HR",          "Kanpur",  46000, 24),
    (9,  "Arjun",  "Sales",       "Lucknow", 52000, 31),
    (10, "Neha",   "Engineering", "Mumbai",  88000, 29),
    (11, "Karan",  "Sales",       "Delhi",   61000, 38),
    (12, "Pooja",  "Marketing",   "Kanpur",  59000, 26),
]

columns = ["emp_id", "name", "department", "city", "salary", "age"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
from pyspark.sql.functions import col,avg,sum
df[col("name"),col("salary")].show()
```

```python
df[col("salary")>60000].show()
```

```python
df[(col("department")=='Engineering') & (col("city")=='Delhi')].show()
```

```python
from pyspark.sql.functions import (
    col, count, sum, avg, min, max,
    row_number, rank, dense_rank,
    round
)
```

```python
df.groupBy("department").count().alias("total_employees").show()
```

```python
df.groupBy("department").agg(
    round(avg(col("salary")),2).alias("average_salary")
).show()
```

```python
df.groupBy("city").count().alias("totol_employees").show()
```

```python
df.groupBy("city").agg(
    count(col("*")).alias("total_employees")
).orderBy(col("total_employees").desc()).show()
```
